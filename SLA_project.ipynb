#TASK 5 – Outlier Handling

```python
cols = ['shipping_cost_usd', 'shipment_weight_kg', 'shipment_volume_cbm', 'delivery_delay_days']
# Dictionary to store thresholds
iqr_thresholds = {}

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    iqr_thresholds[col] = (lower, upper)

    # Count outliers
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col} - outliers: {len(outliers)} | lower: {lower:.2f}, upper: {upper:.2f}")
```

```python
# Cap extreme values where appropriate
# Decision: cap weight and volume, keep shipping cost and delay
cap_cols = ['shipment_weight_kg', 'shipment_volume_cbm']

for col in cap_cols:
    lower, upper = iqr_thresholds[col]
    df[col] = df[col].clip(lower=lower, upper=upper)

print("\nOutlier handling completed.")
```

 TASK 5 – Outlier Handling (IQR Method)

We applied the IQR method to detect and handle outliers in the key numerical columns:

- **shipping_cost_usd**  
  - Some extremely high costs exist (e.g., premium air shipments).  
  - **Decision:** Keep.  
  - **Justification:** These are legitimate business cases, not errors.

- **shipment_weight_kg**  
  - A few shipments are extremely heavy.  
  - **Decision:** Cap at the IQR upper bound.  
  - **Justification:** Prevents rare extreme weights from skewing the model while retaining most data.

- **shipment_volume_cbm**  
  - Some shipments have very large volumes.  
  - **Decision:** Cap at the IQR upper bound.  
  - **Justification:** Large volumes are rare and could distort model predictions; capping keeps the dataset stable.

- **delivery_delay_days**  
  - Some deliveries were delayed by many days.  
  - **Decision:** Keep.  
  - **Justification:** Long delays are real operational events and represent true business risk; removing them would bias the model.
